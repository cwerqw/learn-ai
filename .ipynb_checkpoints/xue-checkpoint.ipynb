```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
t=np.arange(0.0,100.0,1)
s=np.sin(0.1*np.pi*t)*np.exp(-t*0.1)
```

```python
s
```

```python
from matplotlib.ticker import MultipleLocator,FormatStrFormatter
```

```python
#设置  ax中设置 plt里面
ax=plt.subplot(111)
plt.plot(t,s,'--')
plt.grid(True,linestyle='--',color='gray',linewidth='0.5',axis='both')
#将x主刻度标签设置为10的倍数
xmajorLocator=MultipleLocator(10)
#设置x轴文本显示的格式
xmajorFormatter=FormatStrFormatter('%.0f')
#次刻度
xminorLocator=MultipleLocator(5)
ax.xaxis.set_major_locator(xmajorLocator)
ax.xaxis.set_major_formatter(xmajorFormatter)
ax.xaxis.set_minor_locator(xminorLocator)
#ax.yaxis.set_xxxxx
```

```python
import matplotlib.pyplot as plt
import numpy as np

plt.figure(figsize=(22, 10))
x = np.arange(10)
y1 = np.random.rand(10)
y2 = -np.random.rand(10) # 照抄截图，使用负数

# width 宽度比例, facecolor 填充色, yerr 误差棒
# 关键点：使用十六进制颜色，并且 yerr 统一用 np.abs() 取绝对值
plt.bar(x, y1, width=1, facecolor='yellowgreen', edgecolor='white', yerr=np.abs(y1*0.1))
plt.bar(x, y2, width=1, facecolor='#87CEFA', edgecolor='white', yerr=np.abs(y2*0.1))
for i,j in zip(x,y1):
    plt.text(i-0.15,j+0.05,'%.2f'%j,color='red')
```

```python
import pandas as pd#外嵌图表

data = [[ 66386, 174296,  75131, 577908,  32015],
        [ 58230, 381139,  78045,  99308, 160454],
        [ 89135,  80552, 152558, 497981, 603535],
        [ 78415,  81858, 150656, 193263,  69638],
        [139361, 331509, 343164, 781380,  52269]]

columns = ('Freeze', 'Wind', 'Flood', 'Quake', 'Hail')
rows = ['%d year' % x for x in (100, 50, 20, 10, 5)]

df = pd.DataFrame(data, columns=('Freeze', 'Wind', 'Flood', 'Quake', 'Hail'),
                  index=['%d year' % x for x in (100, 50, 20, 10, 5)])

print(df)
```

```python
import matplotlib.pyplot as plt

# 1. 先画柱状图（这一步会确定画布）
df.plot(kind='bar', grid=True, colormap='Blues_r', stacked=True, figsize=(18, 8))

# 2. 再画表格（直接在当前画布上叠加）
plt.table(cellText=data,                         # 表格文本
          cellLoc='center',                      # cell内文本对齐位置
          rowLabels=rows,                        # 行标签
          colColours=None,
          rowColours=plt.cm.PuBu(np.linspace(0, 0.5, 5))[::-1], # 使用 PuBu 渐变
          colLabels=columns,                     # 列标签
         )

# 3. 最后统一显示
plt.show()
```
